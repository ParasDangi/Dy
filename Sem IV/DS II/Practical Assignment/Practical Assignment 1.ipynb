```python
class BTNode:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
 
    def printTree(self):
        if self.left:
            self.left.printTree()
        print(self.data,end=" ")
        if self.right:
            self.right.printTree()

root=BTNode(10)
n2=BTNode(20)
n3=BTNode(30)
n4=BTNode(40)
n5=BTNode(50)
n6=BTNode(60)
n7=BTNode(70)

root.left=n2
root.right=n3
n2.left=n4
n2.right=n5
n3.left=n6
n3.right=n7
root.printTree()
print()
```

```python
class BTNode:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None

def insertLevelwise(a,i,n):
    root=None
    if i<n:
        root=BTNode(a[i])
        root.left=insertLevelwise(a,2*i+1,n)
        root.right=insertLevelwise(a,2*i+2,n)
    return root

def inorder(root):
    if root!=None:
        inorder(root.left)
        print(root.data,end=" ")
        inorder(root.right)

def preorder(root):
    if root!=None:
        print(root.data,end=" ")
        preorder(root.left)
        preorder(root.right)

def postorder(root):
    if root!=None:
        postorder(root.left)
        postorder(root.right)
        print(root.data,end=" ")
        
a=[10,20,30,40,50,60,70]
n=len(a)
root=insertLevelwise(a,0,n)

inorder(root)
print()

preorder(root)
print()

postorder(root)
print()
```
